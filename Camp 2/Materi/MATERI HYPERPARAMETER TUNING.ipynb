# 📘 MATERI HYPERPARAMETER TUNING

## Apa itu Hyperparameter?

Hyperparameter adalah parameter yang ditentukan sebelum proses training dan tidak dipelajari langsung oleh model dari data.

Contoh:
*   Learning rate
*   Jumlah pohon (n_estimators) pada Random Forest
*   Nilai K pada KNN
*   Nilai C dan gamma pada SVM

Berbeda dengan model parameter:
*   Model parameter → dipelajari dari data (weight, bias)
*   Hyperparameter → diatur oleh data scientist

# Jenis-Jenis Hyperparameter
## Hyperparameter Model
| Algoritma           | Hyperparameter               |
| ------------------- | ---------------------------- |
| KNN                 | n_neighbors                  |
| Decision Tree       | max_depth, min_samples_split |
| Random Forest       | n_estimators, max_depth      |
| SVM                 | C, kernel, gamma             |
| Logistic Regression | C, penalty                   |

## Hyperparameter Training
* Learning rate
* Batch size
* Epoch
* Optimizer (pada deep learning)

# Teknik Hyperparameter Tuning
## Manual Search
Menentukan hyperparameter berdasarkan pengalaman.

Kekurangan:

* Tidak efisien

* Tidak sistematis

* Rentan bias

## Grid Search
Mencoba semua kombinasi hyperparameter yang ditentukan.

Kelebihan:

* Sistematis

* Hasil optimal pada ruang pencarian

Kekurangan:

* Lambat

* Komputasi mahal

```python
# Contoh Grid Search (Random Forest)

from sklearn.ensemble import RandomForestClassifier       # Model Random Forest untuk klasifikasi
from sklearn.model_selection import GridSearchCV          # Tools untuk hyperparameter tuning Grid Search
from sklearn.datasets import load_iris                    # Dataset Iris bawaan sklearn

# Load dataset
data = load_iris()                                        # Memuat dataset Iris
X = data.data                                             # Fitur (input)
y = data.target                                           # Label/kelas (output)

# Model
rf = RandomForestClassifier(random_state=42)              # Inisialisasi model dengan seed agar hasil konsisten

# Hyperparameter grid
param_grid = {
    'n_estimators': [50, 100, 200],                       # Jumlah pohon dalam Random Forest
    'max_depth': [None, 5, 10],                           # Kedalaman maksimum tiap pohon
    'min_samples_split': [2, 5]                           # Minimum data untuk memecah node
}

# Grid Search
grid_search = GridSearchCV(
    estimator=rf,                                         # Model yang akan dituning
    param_grid=param_grid,                                # Daftar kombinasi hyperparameter
    cv=5,                                                 # 5-fold Cross Validation
    scoring='accuracy',                                   # Metrik evaluasi yang digunakan
    n_jobs=-1                                             # Gunakan semua core CPU
)

grid_search.fit(X, y)                                     # Melatih semua kombinasi hyperparameter

# Hasil terbaik
print("Best Parameters:", grid_search.best_params_)       # Hyperparameter terbaik
print("Best Accuracy:", grid_search.best_score_)          # Akurasi rata-rata terbaik (CV)
```

# Random Search

Mencoba kombinasi hyperparameter secara acak dalam jumlah iterasi tertentu.

Kelebihan:

* Lebih cepat

* Efektif untuk ruang parameter besar

Kekurangan:

* Tidak menjamin kombinasi terbaik dicoba

```python
# Contoh Random Search

from sklearn.model_selection import RandomizedSearchCV     # Tools hyperparameter tuning Random Search
from scipy.stats import randint                            # Distribusi bilangan bulat acak

param_dist = {
    'n_estimators': randint(50, 300),                      # Jumlah pohon dipilih acak antara 50–300
    'max_depth': randint(3, 15),                           # Kedalaman pohon dipilih acak antara 3–15
    'min_samples_split': randint(2, 10)                    # Minimum data untuk split node (acak)
}

random_search = RandomizedSearchCV(
    estimator=rf,                                          # Model yang akan dituning
    param_distributions=param_dist,                        # Distribusi hyperparameter
    n_iter=20,                                             # Jumlah kombinasi acak yang dicoba
    cv=5,                                                  # 5-fold Cross Validation
    scoring='accuracy',                                    # Metrik evaluasi
    random_state=42,                                       # Seed agar hasil konsisten
    n_jobs=-1                                              # Gunakan semua core CPU
)

random_search.fit(X, y)                                    # Melatih model dengan kombinasi acak

print("Best Parameters:", random_search.best_params_)      # Hyperparameter terbaik
print("Best Accuracy:", random_search.best_score_)         # Akurasi rata-rata terbaik (CV)
```
